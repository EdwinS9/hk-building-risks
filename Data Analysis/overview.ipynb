```python
import pandas as pd

df = pd.read_csv("DS1_all.csv")
df.head()
```

```python
import plotly.express as px

df["completion_date"] = pd.to_datetime(df["NSEARCH3_E"], errors="coerce")
df["age"] = (pd.Timestamp.today() - df["completion_date"]).dt.days / 365.25

fig = px.histogram(df, x="age", nbins=50, title="Distribution of Building Age",
                   labels={"age": "Building Age (years)"})
fig.show()
```

```python
import plotly.graph_objects as go
import pandas as pd

ds2 = pd.read_csv("DS2_inspection.csv")
ds2_age = ds2.merge(df[["OBJECTID", "age"]], on="OBJECTID", how="inner")["age"].dropna()
ds1_age = df["age"].dropna()

fig = go.Figure()
for ages, name, color in [
    (ds1_age, "DS1 – all buildings", "steelblue"),
    (ds2_age, "DS2 – noticed buildings", "tomato"),
]:
    fig.add_trace(go.Histogram(
        x=ages, nbinsx=50, name=name,
        histnorm="probability density",
        marker_color=color, opacity=0.6,
    ))

fig.update_layout(
    barmode="overlay",
    title="Age distribution: DS1 vs DS2 (density-normalised)",
    xaxis_title="Building Age (years)",
    yaxis_title="Density",
    legend=dict(x=0.01, y=0.99),
)
fig.show()
```
