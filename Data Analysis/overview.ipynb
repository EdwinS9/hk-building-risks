```python
import pandas as pd

df = pd.read_csv("DS1_all.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

df["completion_date"] = pd.to_datetime(df["NSEARCH3_E"], errors="coerce")
df["age"] = (pd.Timestamp.today() - df["completion_date"]).dt.days / 365.25

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df["age"].dropna(), bins=50, edgecolor="white", linewidth=0.4)
ax.set_xlabel("Building Age (years)")
ax.set_ylabel("Count")
ax.set_title("Distribution of Building Age")
plt.tight_layout()
plt.show()
```
